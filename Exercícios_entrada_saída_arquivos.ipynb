```python
#Lendo e escrevendo ao mesmo tempo

arquivo_entrada = 'gene KRT18.fna'
arquivo_saida = 'teste_saida.fasta'

with open(arquivo_entrada, 'r') as f, open(arquivo_saida, 'w') as g:
    for linha in f:
        transcrito = linha.replace('T','U')
        # print(linha)
        # print(transcrito)
        g.write(transcrito)
```

```python
#os.path.exists

#Verifica que se o diretório existe no caminho especificado
#Se não existir o arquivo um novo será criado com o texto "NOVA LINHA"

arquivo = "gene14.fasta"

if os.path.exists(arquivo) == True:
    print(f"O arquivo {arquivo} existe.")
    modo = 'a'
else:
    print(f"O arquivo {arquivo} não foi encontrado.")
    modo = 'w'

with open(arquivo, modo) as f:
    print(f"Arquivo {arquivo} aberto em modo '{modo}'.")
    f.write('NOVA LINHA\n')
```
