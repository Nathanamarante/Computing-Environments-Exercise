### **Listas de exercícios 5 - Variáveis - conjuntos**

```python
#1

conj1 = set ([1.9, 1.8, 5.7, 1.6, 5.8, 1.7, 9.6, 5.9, 9.5, 6.5, 6.2, 1.1, 4.4, 3.5, 2.9, 4.7,
4.6, 5.2, 5.3])
conj2 = set ([4.7, 3.6, 6.2, 7.1, 7, 5.6, 5.7, 3.4, 3.3, 2.1, 3.8, 3.9, 5, 5.1, 1.9, 9.5,
1.0, 1.3, 5.4])
conj3 = set ([2.2, 3.3, 5.1, 3, 3.7, 9.1, 8.8, 8.5, 2, 4.1, 6.1, 4.9, 1.1, 0.5, 0.8, 3.2,
6.9, 9.3, 9.5])

dif1_2 = conj1 - conj2
print(f'diferença conjunto 1 e 2: {dif1_2}') #Descobri o uso das F strings

dif1_3 = conj1 - conj3
print(f'diferença conjunto 1 e 23: {dif1_3}')

dif2_3 = conj2 - conj3
print(f'diferença conjunto 2 e 3: {dif2_3}')


intersec1_2 = conj1.intersection(conj2)
print(f'interseção conjunto 1 e 2: {intersec1_2}')

intersec1_3 = conj1.intersection(conj3)
print(f'interseção conjunto 1 e 2: {intersec1_3}')

intersec2_3 = conj2.intersection(conj3)
print(f'interseção conjunto 1 e 2: {intersec2_3}')

insercao1 = conj1.union(conj2)
insercao2= conj1.union(conj3)

insercao3 = insercao1.union(insercao2)
print(f'União elementos 2 e 3 em 1: {insercao3}')

print(f'Tamanho conjunto tarefa C: {len(insercao3)}')
```

```python
#2

A = {3, 6, 9, 12, 15, 18, 21, 24, 28, 27}
B = {2, 6, 8, 10, 14, 16, 18, 20, 22, 24}
C = {2, 6, 10, 18, 20}
D = {1, 30, 5, 11, 17, 16, 22, 26}

intersecaoA_B = A.intersection(B)
print(f"Interseção de A e B: {intersecaoA_B}")

diferencaA_B = A - B
print(f"Diferença de A e B (A - B): {diferencaA_B}")


intersecaoA_D = A.intersection(D)
A_D_disjunto = len(intersecaoA_D) == 0
print(f"A e D são disjuntos?: {A_D_disjunto}")
print(f"Interseção de A e D: {intersecaoA_D}")


intersecaoB_D = B.intersection(D)
B_D_disjunto = len(intersecaoB_D) == 0
print(f"B e D são disjuntos?: {B_D_disjunto}")
print(f"Interseção de B e D: {intersecaoB_D}")

C_subconjunto_A = C.issubset(A)
print(f'C é subconjunto de A: {C_subconjunto_A}')

C_subconjunto_B = C.issubset(B)
print(f'C é subconjunto de B:{C_subconjunto_B}')

D.update({18, 23, 25, 63})
print(f'Inserção de 18, 23, 25, 63 em D: {D}')
```
