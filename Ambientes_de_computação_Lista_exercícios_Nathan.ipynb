## **Lista de exercícios 7 - Variáveis - Operadores**

```python
#1

import math

Glicina1 = {
            "N": (108.304, 100.827, 67.992),
            "CA": (108.477, 100.389, 69.362),
            "C": (109.907, 100.555, 69.817),
            "O": (110.821, 100.799, 69.027)
            }

Glicina2 = {
            "N": (107.670, 101.359, 70.074),
            "CA": (108.477, 100.389, 69.362),
            "C": (109.513, 101.011, 68.450),
            "O": (110.667, 100.572, 68.425)
            }

print(f'Valores Glicina1: {Glicina1}')
print(f'Valores Glicina2: {Glicina2}')

def rmsd(estrutura1, estrutura2, N=4):
    soma = 0.0
    for atomo in estrutura1:
        x1, y1, z1 = estrutura1[atomo]
        x2, y2, z2 = estrutura2[atomo]
        soma += (x1 - x2)**2 + (y1 - y2)**2 + (z1 - z2)**2
    return math.sqrt(soma / N)


valor_rmsd = rmsd(Glicina1, Glicina2, N=4)
print(f'RMSD = {valor_rmsd:.4f}')
```
